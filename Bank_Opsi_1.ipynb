```python

import ipywidgets as widgets
from IPython.display import display, HTML, clear_output
from html import escape

# ==========================================
# 1. CLASS DAN STRUKTUR DATA
# ==========================================

class Nasabah:
    def __init__(self, nomor_antrean, nama, status, keperluan):
        self.nomor_antrean = nomor_antrean
        self.nama = nama
        self.status = status.upper()
        self.prioritas = 1 if self.status == "VIP" else 2
        self.keperluan = keperluan

    def __str__(self):
        return f"[{self.nomor_antrean}] {self.nama} ({self.status}) - {self.keperluan}"


class Node:
    def __init__(self, nasabah):
        self.nasabah = nasabah
        self.next = None


class PriorityQueueBank:
    def __init__(self):
        self.head = None
        self.size = 0

    def is_empty(self):
        return self.head is None

    def enqueue(self, nasabah):
        new_node = Node(nasabah)

        if self.is_empty() or nasabah.prioritas < self.head.nasabah.prioritas:
            new_node.next = self.head
            self.head = new_node
        else:
            current = self.head

            while (
                current.next is not None
                and current.next.nasabah.prioritas <= nasabah.prioritas
            ):
                current = current.next

            new_node.next = current.next
            current.next = new_node

        self.size += 1

    def dequeue(self):
        if self.is_empty():
            return None

        nasabah = self.head.nasabah
        self.head = self.head.next
        self.size -= 1

        return nasabah

    def get_all_items(self):
        items = []
        current = self.head

        while current is not None:
            items.append(current.nasabah)
            current = current.next

        return items


# ==========================================
# 2. INISIALISASI SISTEM
# ==========================================

queue = PriorityQueueBank()

counter_vip = 1
counter_reg = 1

total_dilayani = 0
total_masuk = 0

sedang_dilayani = None
notifikasi = "Sistem antrean berhasil dijalankan."


data_awal = [
    ("Budi Santoso", "REGULER", "Setor / Tarik Tunai"),
    ("Siti Aminah", "REGULER", "Buka Rekening"),
    ("Andi Wijaya", "VIP", "Kredit & Investasi"),
    ("Rina Kartika", "REGULER", "Customer Service"),
    ("Dewi Anggraini", "VIP", "Buka Rekening")
]

for nama, status, keperluan in data_awal:

    if status == "VIP":
        nomor = f"V-{counter_vip:03d}"
        counter_vip += 1
    else:
        nomor = f"R-{counter_reg:03d}"
        counter_reg += 1

    queue.enqueue(Nasabah(nomor, nama, status, keperluan))
    total_masuk += 1


# ==========================================
# 3. CSS ANIMATION DAN DESIGN
# ==========================================

display(HTML("""
<style>

@import url('https://fonts.googleapis.com/css2?family=Inter:wght@400;500;600;700;800&display=swap');

.widget-label {
    font-family: Inter, sans-serif !important;
}

.bank-container {
    font-family: Inter, sans-serif;
    color: #1e293b;
    animation: fadeIn 0.7s ease;
}

@keyframes fadeIn {
    from {opacity: 0; transform: translateY(15px);}
    to {opacity: 1; transform: translateY(0);}
}

@keyframes slideIn {
    from {
        opacity: 0;
        transform: translateX(-25px);
    }
    to {
        opacity: 1;
        transform: translateX(0);
    }
}

@keyframes pulse {
    0%, 100% {box-shadow: 0 0 0 0 rgba(16,185,129,0.25);}
    50% {box-shadow: 0 0 0 8px rgba(16,185,129,0);}
}

@keyframes glow {
    0%, 100% {box-shadow: 0 0 5px rgba(245,158,11,0.1);}
    50% {box-shadow: 0 0 15px rgba(245,158,11,0.4);}
}

@keyframes ticker {
    0% {opacity: 0.4; transform: scale(0.98);}
    100% {opacity: 1; transform: scale(1);}
}

.bank-header {
    background: linear-gradient(120deg, #0f172a, #1e40af, #2563eb);
    padding: 30px;
    border-radius: 18px;
    color: white;
    position: relative;
    overflow: hidden;
    box-shadow: 0 10px 30px rgba(30,64,175,0.2);
}

.bank-header:after {
    content: "";
    position: absolute;
    width: 230px;
    height: 230px;
    background: rgba(255,255,255,0.07);
    border-radius: 50%;
    right: -60px;
    top: -100px;
}

.bank-header h2 {
    margin: 0;
    font-size: 25px;
    font-weight: 800;
    letter-spacing: 1px;
}

.bank-header p {
    color: #bfdbfe;
    margin: 9px 0 0;
    font-size: 13px;
}

.stat-card {
    background: white;
    border: 1px solid #e2e8f0;
    border-radius: 14px;
    padding: 18px;
    flex: 1;
    min-width: 120px;
    box-shadow: 0 4px 12px rgba(15,23,42,0.04);
    transition: 0.3s;
}

.stat-card:hover {
    transform: translateY(-5px);
    box-shadow: 0 10px 20px rgba(15,23,42,0.09);
}

.stat-number {
    font-size: 28px;
    font-weight: 800;
    margin-top: 8px;
}

.stat-label {
    font-size: 11px;
    font-weight: 700;
    color: #64748b;
    text-transform: uppercase;
    letter-spacing: 0.7px;
}

.teller-card {
    background: linear-gradient(110deg, #ecfdf5, #f0fdf4);
    border: 1px solid #a7f3d0;
    border-radius: 15px;
    padding: 22px;
    animation: ticker 0.5s ease;
}

.teller-name {
    color: #047857;
    font-size: 20px;
    font-weight: 800;
    margin-top: 8px;
}

.queue-card {
    display: flex;
    justify-content: space-between;
    align-items: center;
    gap: 10px;
    padding: 16px;
    border-radius: 12px;
    margin-bottom: 10px;
    animation: slideIn 0.5s ease both;
    transition: 0.3s;
}

.queue-card:hover {
    transform: translateX(5px);
}

.vip-card {
    background: linear-gradient(100deg, #fffbeb, #fef3c7);
    border: 1px solid #fcd34d;
    animation: slideIn 0.5s ease both, glow 3s infinite;
}

.reg-card {
    background: #f8fafc;
    border: 1px solid #e2e8f0;
}

.queue-number {
    font-size: 12px;
    color: #64748b;
    margin-bottom: 5px;
}

.queue-name {
    font-size: 15px;
    font-weight: 700;
    color: #1e293b;
}

.queue-detail {
    font-size: 12px;
    color: #64748b;
    margin-top: 4px;
}

.vip-badge {
    background: #d97706;
    color: white;
    padding: 6px 10px;
    border-radius: 20px;
    font-size: 10px;
    font-weight: 800;
    white-space: nowrap;
}

.reg-badge {
    background: #e2e8f0;
    color: #475569;
    padding: 6px 10px;
    border-radius: 20px;
    font-size: 10px;
    font-weight: 800;
    white-space: nowrap;
}

.section-card {
    background: white;
    border: 1px solid #e2e8f0;
    padding: 22px;
    border-radius: 16px;
    box-shadow: 0 5px 15px rgba(15,23,42,0.035);
}

.section-title {
    font-size: 16px;
    font-weight: 800;
    color: #0f172a;
    padding-bottom: 14px;
    border-bottom: 1px solid #e2e8f0;
    margin-bottom: 18px;
}

.progress-bg {
    height: 8px;
    border-radius: 20px;
    background: #e2e8f0;
    overflow: hidden;
    margin-top: 12px;
}

.progress-fill {
    height: 100%;
    background: linear-gradient(90deg, #2563eb, #60a5fa);
    border-radius: 20px;
    transition: width 0.7s ease;
}

.notice {
    background: #eff6ff;
    border-left: 4px solid #3b82f6;
    padding: 13px;
    border-radius: 8px;
    color: #1d4ed8;
    font-size: 12px;
    margin-top: 15px;
    animation: fadeIn 0.4s ease;
}

.empty-state {
    text-align: center;
    padding: 35px 10px;
    color: #94a3b8;
}

.footer-bank {
    text-align: center;
    color: #94a3b8;
    font-size: 11px;
    padding: 20px;
}

</style>
"""))


# ==========================================
# 4. KOMPONEN INPUT GUI
# ==========================================

output_area = widgets.Output()

txt_nama = widgets.Text(
    description="Nama:",
    placeholder="Masukkan nama nasabah",
    layout=widgets.Layout(width="100%")
)

cmb_keperluan = widgets.Dropdown(
    options=[
        "Setor / Tarik Tunai",
        "Buka Rekening",
        "Kredit & Investasi",
        "Customer Service"
    ],
    description="Keperluan:",
    layout=widgets.Layout(width="100%")
)

rdo_kategori = widgets.RadioButtons(
    options=[
        ("REGULER", "REGULER"),
        ("VIP PRIORITAS", "VIP")
    ],
    value="REGULER",
    description="Kategori:",
    layout=widgets.Layout(width="100%")
)

btn_tambah = widgets.Button(
    description="AMBIL NOMOR ANTREAN",
    icon="plus-circle",
    button_style="primary",
    layout=widgets.Layout(width="100%", height="45px")
)

btn_panggil = widgets.Button(
    description="PANGGIL NASABAH BERIKUTNYA",
    icon="bell",
    button_style="success",
    layout=widgets.Layout(width="100%", height="48px")
)


# ==========================================
# 5. FUNGSI RENDER DASHBOARD
# ==========================================

def render_ui():

    with output_area:
        clear_output(wait=True)

        items = queue.get_all_items()

        jumlah_vip = sum(1 for n in items if n.status == "VIP")
        jumlah_reg = sum(1 for n in items if n.status == "REGULER")

        # Header
        header = """
        <div class="bank-container">

            <div class="bank-header">
                <div style="font-size:11px; font-weight:700; color:#93c5fd;">
                    BANK MANAGEMENT SYSTEM
                </div>
                <h2>🏦 BANK QUEUE SYSTEM</h2>
                <p>Smart Priority Queue Management Dashboard</p>
                <div style="margin-top:20px; font-size:11px; color:#dbeafe;">
                    ● SYSTEM ONLINE &nbsp; | &nbsp; SINGLY LINKED LIST
                </div>
            </div>

        """

        # Statistik
        stats = f"""
        <div style="display:flex; flex-wrap:wrap; gap:12px; margin:20px 0;">

            <div class="stat-card">
                <div class="stat-label">🕒 Menunggu</div>
                <div class="stat-number" style="color:#2563eb;">{queue.size}</div>
                <div style="font-size:11px;color:#94a3b8;">Nasabah antre</div>
            </div>

            <div class="stat-card">
                <div class="stat-label">👑 VIP</div>
                <div class="stat-number" style="color:#d97706;">{jumlah_vip}</div>
                <div style="font-size:11px;color:#94a3b8;">Prioritas 1</div>
            </div>

            <div class="stat-card">
                <div class="stat-label">👤 Reguler</div>
                <div class="stat-number" style="color:#64748b;">{jumlah_reg}</div>
                <div style="font-size:11px;color:#94a3b8;">Prioritas 2</div>
            </div>

            <div class="stat-card">
                <div class="stat-label">✅ Dilayani</div>
                <div class="stat-number" style="color:#059669;">{total_dilayani}</div>
                <div style="font-size:11px;color:#94a3b8;">Total selesai</div>
            </div>

        </div>
        """

        # Teller
        if sedang_dilayani is None:
            teller_text = "Belum ada nasabah yang dipanggil"
            teller_nomor = "MENUNGGU PANGGILAN"
        else:
            teller_text = escape(sedang_dilayani.nama)
            teller_nomor = escape(sedang_dilayani.nomor_antrean)

        teller = f"""
        <div class="teller-card" style="margin-bottom:20px; animation:pulse 3s infinite;">

            <div style="font-size:11px; font-weight:800; color:#059669; letter-spacing:1px;">
                ● TELLER 01 - STATUS LAYANAN
            </div>

            <div style="font-size:30px; margin-top:12px;">🔊</div>

            <div class="teller-name">{teller_text}</div>

            <div style="font-size:12px; color:#047857; margin-top:5px;">
                Nomor Antrean: <b>{teller_nomor}</b>
            </div>

        </div>
        """

        # Daftar antrean
        queue_html = ""

        if items:

            for idx, nasabah in enumerate(items, start=1):

                if nasabah.status == "VIP":
                    card_class = "vip-card"
                    badge = '<span class="vip-badge">👑 VIP PRIORITY</span>'
                    icon = "🔥"
                else:
                    card_class = "reg-card"
                    badge = '<span class="reg-badge">REGULER</span>'
                    icon = "👤"

                queue_html += f"""
                <div class="queue-card {card_class}"
                     style="animation-delay:{idx * 0.07}s;">

                    <div>
                        <div class="queue-number">
                            {icon} URUTAN #{idx} &nbsp; • &nbsp;
                            {escape(nasabah.nomor_antrean)}
                        </div>

                        <div class="queue-name">
                            {escape(nasabah.nama)}
                        </div>

                        <div class="queue-detail">
                            📋 {escape(nasabah.keperluan)}
                        </div>
                    </div>

                    <div>{badge}</div>

                </div>
                """

        else:

            queue_html = """
            <div class="empty-state">
                <div style="font-size:45px;">📭</div>
                <h3 style="color:#64748b;">Antrean Kosong</h3>
                <p>Semua nasabah telah mendapatkan layanan.</p>
            </div>
            """

        # Progress
        progress = min((total_dilayani / max(total_masuk, 1)) * 100, 100)

        dashboard = f"""
        <div class="section-card" style="margin-bottom:20px;">

            <div class="section-title">
                📋 DAFTAR ANTREAN NASABAH
            </div>

            {queue_html}

            <div style="display:flex; justify-content:space-between;
                        margin-top:20px; font-size:12px; color:#64748b;">

                <span>Total nasabah menunggu</span>
                <b style="color:#2563eb;">{queue.size} Nasabah</b>

            </div>

            <div class="progress-bg">
                <div class="progress-fill" style="width:{progress}%;"></div>
            </div>

            <div style="font-size:10px; color:#94a3b8; margin-top:7px;">
                Progres pelayanan: {progress:.1f}%
            </div>

        </div>

        <div class="notice">
            🔔 <b>Notifikasi Sistem:</b> {escape(notifikasi)}
        </div>

        <div class="footer-bank">
            BANK QUEUE MANAGEMENT SYSTEM © 2026
            <br>
            Powered by Python • Priority Queue • Singly Linked List
        </div>

        </div>
        """

        display(HTML(header + stats + teller + dashboard))


# ==========================================
# 6. EVENT HANDLER
# ==========================================

def on_tambah_clicked(b):

    global counter_vip, counter_reg, total_masuk, notifikasi

    nama = txt_nama.value.strip()

    if not nama:
        notifikasi = "Silakan masukkan nama nasabah terlebih dahulu."
        render_ui()
        return

    status = rdo_kategori.value
    keperluan = cmb_keperluan.value

    if status == "VIP":
        nomor = f"V-{counter_vip:03d}"
        counter_vip += 1
    else:
        nomor = f"R-{counter_reg:03d}"
        counter_reg += 1

    nasabah = Nasabah(nomor, nama, status, keperluan)

    queue.enqueue(nasabah)

    total_masuk += 1

    notifikasi = f"Nasabah {nama} berhasil mendapatkan nomor {nomor}."

    txt_nama.value = ""

    render_ui()


def on_panggil_clicked(b):

    global sedang_dilayani, total_dilayani, notifikasi

    nasabah = queue.dequeue()

    if nasabah is None:

        sedang_dilayani = None
        notifikasi = "Tidak ada nasabah dalam antrean."

    else:

        sedang_dilayani = nasabah
        total_dilayani += 1

        notifikasi = (
            f"Nasabah {nasabah.nama} dengan nomor "
            f"{nasabah.nomor_antrean} sedang dilayani."
        )

    render_ui()


# ==========================================
# 7. HUBUNGKAN TOMBOL
# ==========================================

btn_tambah.on_click(on_tambah_clicked)
btn_panggil.on_click(on_panggil_clicked)


# ==========================================
# 8. LAYOUT APLIKASI
# ==========================================

form_box = widgets.VBox(
    [
        widgets.HTML("""
        <h3 style="color:#1e293b; margin:0;">
            📝 Loket Pendaftaran
        </h3>
        <p style="font-size:12px;color:#94a3b8;">
            Masukkan data nasabah baru
        </p>
        """),
        txt_nama,
        cmb_keperluan,
        rdo_kategori,
        btn_tambah
    ],
    layout=widgets.Layout(
        width="65%",
        padding="20px",
        border="1px solid #e2e8f0",
        border_radius="14px"
    )
)

action_box = widgets.VBox(
    [
        widgets.HTML("""
        <h3 style="color:#1e293b; margin:0;">
            🏧 Operasional Teller
        </h3>
        <p style="font-size:12px;color:#94a3b8;">
            Kontrol pemanggilan antrean
        </p>
        """),
        widgets.HTML("""
        <div style="background:#eff6ff;padding:15px;border-radius:10px;
                    color:#1d4ed8;font-size:12px;margin:5px 0 15px;">
            <b>Priority Queue Aktif</b><br>
            VIP akan dipanggil terlebih dahulu dibandingkan Reguler.
        </div>
        """),
        btn_panggil
    ],
    layout=widgets.Layout(
        width="35%",
        padding="20px",
        border="1px solid #e2e8f0",
        border_radius="14px"
    )
)

layout_input = widgets.HBox(
    [form_box, action_box],
    layout=widgets.Layout(
        width="100%",
        gap="15px",
        align_items="stretch"
    )
)


# ==========================================
# 9. TAMPILKAN APLIKASI
# ==========================================

display(layout_input)
display(output_area)

render_ui()
```
